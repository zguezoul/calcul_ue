```python
#version 4: 





# ************** VERSION 4 **********************

import matplotlib.pyplot as plt

matieres = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]

list_notes = [i for i in range(21)]

ue1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
ue2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
ue3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]


# Création des dictionnaires

def crea_dico(liste):
    dico = {cle : valeur for cle, valeur in zip(matieres,liste)}
    return dico


dico_notes = crea_dico(list_notes)
dico_coef1 = crea_dico(ue1)
dico_coef2 = crea_dico(ue2)
dico_coef3 = crea_dico(ue3)


# Calcul des moyennes

def calcul_moy(dico):
    somme = 0
    total_coef = 0

    for matiere in matieres:
        somme += dico_notes[matiere] * dico[matiere]
        total_coef += dico[matiere]

    moy = round(somme / total_coef, 3)

    return moy


moy1 = calcul_moy(dico_coef1)
moy2 = calcul_moy(dico_coef2)
moy3 = calcul_moy(dico_coef3)


print("Moyenne UE1 :", moy1)
print("Moyenne UE2 :", moy2)
print("Moyenne UE3 :", moy3)


# Décision du jury

moyennes = [moy1, moy2, moy3]

nb_validees = 0

for moy in moyennes:
    if moy >= 10:
        nb_validees += 1


if nb_validees == 3:
    print("Année validée ! Passage en BUT 2 autorisé.")

elif nb_validees == 2:

    for moy in moyennes:
        if moy < 10:
            if moy >= 8:
                print("Passage en BUT 2 autorisé.")
                print("Il faudra valider cette UE en BUT 2.")
            else:
                print("Année non validée.")
                print("Passage en BUT 2 refusé.")

else:
    print("Année non validée.")
    print("Passage en BUT 2 exclu.")


# Visualisation graphique

plt.bar(["UE1", "UE2", "UE3"], moyennes)

plt.axhline(y=10, linestyle="--", label="Moyenne 10")
plt.axhline(y=8, linestyle=":", label="Moyenne 8")

plt.title("Moyennes des trois UE")
plt.xlabel("Unités d'enseignement")
plt.ylabel("Moyennes")

plt.ylim(0, 20)
plt.legend()
plt.show()


```

```python

# ************** VERSION 5 **********************

from tkinter import *
from tkinter import ttk

fenetre = Tk()

# Taille de la fenêtre
fenetre.geometry("800x600")

frm = ttk.Frame(fenetre)
frm.grid()

ress = [f"R{i}" for i in range(101, 116)]
sae = [f"SAE{i}" for i in range(11, 17)]

notes = []

def notes_(c, liste):

    for i, matiere in enumerate(liste):

        etiq = ttk.Label(frm, text=matiere)
        etiq.grid(column=c, row=i, padx=10, pady=10)

        saisie = ttk.Entry(frm)
        saisie.grid(column=c+1, row=i, padx=10, pady=10)

        notes.append(saisie)


notes_(0, ress)
notes_(2, sae)


# Récupérer les notes saisies
def recup():

    saisies = [i.get() for i in notes]

    print(saisies)


btn1 = ttk.Button(frm, text="Valider", command=recup)
btn1.grid(column=3, row=16)


fenetre.mainloop()
```
